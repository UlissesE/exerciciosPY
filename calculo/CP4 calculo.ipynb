## Ex01 (1,0 pontos) – A Tabela abaixo fornece a população do mundo no século 20.



a. Desenhe um gráfico de dispersão (scatter plot) e observe que um modelo linear não é apropriado.

b. Encontre uma função que modele o crescimento populacional.

c. Desenhe um gráfico da função que você encontrou junto com o gráfico de dispersão. Quão bem o modelo se ajusta aos dados?

d. Use o modelo que você encontrou para prever a população mundial no ano 2020.

```python
# Exemplo - pode ser usado no Ex02!
import numpy as np
import matplotlib.pyplot as plt

# dados
anos = np.array([1900,1910,1920,1930,1940,1950,1960,1970,1980,1990,2000], dtype=float)
pop  = np.array([1.65,1.75,1.86,2.07,2.30,2.52,3.02,3.70,4.45,5.28,6.12], dtype=float)

# centralizar o ano deixa os números menores e o ajuste mais estável
ano0 = anos.min()
X = anos - ano0

# ajuste quadrático: pop ≈ c0 + c1*X + c2*X^2
c2, c1, c0 = np.polyfit(X, pop, 2)

# função para prever
def f(ano):
    x = np.asarray(ano, dtype=float) - ano0
    return c0 + c1*x + c2*(x**2)

# qualidade do ajuste (R²)
y_hat = f(anos)
ss_res = np.sum((pop - y_hat)**2)
ss_tot = np.sum((pop - pop.mean())**2)
r2 = 1 - ss_res/ss_tot
print(f"Modelo: y = {c0:.4f} + {c1:.6f}·(ano-{int(ano0)}) + {c2:.9f}·(ano-{int(ano0)})^2")
print(f"R² = {r2:.4f}")

# previsão para 2020
pred_2020 = f(2020)
print(f"Previsão para 2020: {pred_2020:.2f} bilhões")

plt.scatter(anos, pop, alpha=0.7, edgecolors='black', linewidth=1.5)
```

```python
import numpy as np
import matplotlib.pyplot as plt
 
# dados
anos = np.array([1900, 1910, 1920, 1930, 1940, 1950, 1960, 1970, 1980, 1990, 2000])
pop = np.array([1.65, 1.75, 1.86, 2.07, 2.30, 2.52, 3.02, 3.70, 4.45, 5.28, 6.12])
 
# centralizar o ano deixa os números menores e o ajuste mais estável
ano0 = anos.min()
X = anos - ano0

# ajuste quadrático
coeffs = np.polyfit(X, pop, 2)
c0, c1, c2 = coeffs
 
print(f"b. A função que modela o crescimento é uma função quadrática:")
print(f"   P(t) = {c0:.4f}*(t - {ano0})² + {c1:.4f}*(t - {ano0}) + {c2:.4f}")
 
# C. Desenhar o gráfico da função e avaliar o ajuste
X_model = np.linspace(X.min(), X.max(), 100)
pop_model = np.polyval(coeffs, X_model)
 
plt.figure(figsize=(10, 6))
plt.scatter(anos, pop, color='blue', label='Dados da Tabela')
plt.plot(X_model + ano0, pop_model, color='red', label='Modelo Quadrático')
plt.title('Ajuste do Modelo Quadrático aos Dados de População')
plt.xlabel('Ano')
plt.ylabel('População (em bilhões)')
plt.grid(True)
plt.legend()
plt.show()
 
# Calculando a qualidade do ajuste (R²)
pop_hat = np.polyval(coeffs, X)
ss_res = np.sum((pop - pop_hat)**2)
ss_tot = np.sum((pop - pop.mean())**2)
r2 = 1 - (ss_res / ss_tot)
 
print(f"c. O modelo quadrático se ajusta aos dados.")
print(f"   A qualidade do ajuste, medida pelo R², é de {r2:.4f}.")
print("   Um valor de R² próximo de 1 indica um ajuste excelente.")
 
# D. Prever a população mundial no ano 2020
ano_previsao = 2020
X_previsao = ano_previsao - ano0
pop_previsao = np.polyval(coeffs, X_previsao)
 
print("d. Prevendo a população mundial em 2020:")
print(f"   População prevista para 2020: {pop_previsao:.2f} bilhões de pessoas")
```

```python
# sua solução (célula 1)
import numpy as np
import matplotlib.pyplot as plt
 
# Dados
distancia_d = np.array([0.387, 0.723, 1.000, 1.523, 5.203, 9.541, 19.190, 30.086, 39.507])
periodo_T = np.array([0.241, 0.615, 1.000, 1.881, 11.861, 29.457, 84.008, 164.784, 248.350])
 
# B:
log_distancia = np.log(distancia_d)
log_periodo = np.log(periodo_T)
 
coeficientes = np.polyfit(log_distancia, log_periodo, 1)
exponente = coeficientes[0]
log_a = coeficientes[1]
a = np.exp(log_a)
 
distancia_modelo = np.linspace(0.1, 40, 100)
periodo_modelo = a * distancia_modelo**exponente
 
print(f"Função: a * distancia_modelo**exponente")
print(f"Onde a = {a:.6f} e exponente = {exponente:.6f}")
 
# C:
plt.figure(figsize=(10, 6))
plt.scatter(distancia_d, periodo_T, color='red', label='Dados dos Planetas')
plt.plot(distancia_modelo, periodo_modelo, color='blue', label='Modelo de Potência')
plt.title('Ajuste do Modelo de Potência aos Dados Planetários')
plt.xlabel('Distância (AU)')
plt.ylabel('Período (anos)')
plt.grid(True)
plt.legend()
plt.show()
 
y_log = exponente * log_distancia + log_a
ss_total = np.sum((log_periodo - np.mean(log_periodo))**2)
ss_residual = np.sum((log_periodo - y_log)**2)
r_squared = 1 - (ss_residual / ss_total)
 
print(f'R² = {r_squared:.6f}')
 
# D:
distance_asteroid = 5
period_asteroid = a * distance_asteroid**exponente
 
print(f'O período de um asteroide a {distance_asteroid} AU do sol é de {period_asteroid:.2f} anos.')
 
```

## Ex03 (6,0 pontos) – Encontre a equação da reta tangente e secante à  no ponto .

a. Encontre a equação da reta tangente e normal à $y = x^2-e^x$ no ponto $P = (0, -1)$.

b. Represente a soluções graficamente.

```python
# sua solução (célula 1)

# Equações:
print("Equação da Tangente: y = -x - 1")
print("Equação da Normal:   y =  x - 1")
```

```python
# sua solução (célula 2)
import numpy as np
import matplotlib.pyplot as plt
 
# Definição da função
def f(x):
    return x**2 - np.exp(x)
 
# Derivada da função
def f_prime(x):
    return 2*x - np.exp(x)
 
# Ponto dado
x0 = 0
y0 = f(x0)
 
# Inclinação da tangente e normal
m_tan = f_prime(x0)   # -1
m_norm = -1/m_tan     # 1
 
# Equações da reta tangente e normal
# y - y0 = m*(x - x0)
def tangent(x):
    return m_tan*(x - x0) + y0
 
def normal(x):
    return m_norm*(x - x0) + y0
 
# Gráficos
xs = np.linspace(-3, 3, 400)
plt.figure(figsize=(7,5))
plt.plot(xs, f(xs), label='y = x^2 - e^x', color='blue')
plt.plot(xs, tangent(xs), label='Tangente: y=-x-1', color='orange')
plt.plot(xs, normal(xs), label='Normal: y=x-1', color='green')
plt.scatter([x0], [y0], color='red', zorder=5, label='P (0,-1)')
plt.title('Função, Tangente e Normal no ponto P=(0,-1)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```
