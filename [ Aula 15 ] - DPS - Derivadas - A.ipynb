# **Conceitos**

* **Regra da cadeia** (função composta): se $y=f(g(x))$, então

  $$
  \frac{dy}{dx}=f'(g(x))\cdot g'(x).
  $$

  Útil quando há “camadas” (ex.: $\sigma(wx+b)$, exponenciais dentro de outras funções).

* **Regra do produto**: se $h(x)=u(x)\,v(x)$, então

  $$
  h'(x)=u'(x)v(x)+u(x)v'(x).
  $$

  Aparece quando uma grandeza é produto de duas funções que variam.

* **Regra do quociente**: se $h(x)=\dfrac{u(x)}{v(x)}$, então

  $$
  h'(x)=\frac{u'(x)v(x)-u(x)v'(x)}{[v(x)]^2}.
  $$

  Use quando há divisão de funções (ex.: “custo por usuário”, “latência média” etc.).

**Dicas:** Exponenciais $e^{g(x)}$ pedem cadeia $(e^{g})' = e^{g}\,g'$; potências $(g(x))^n$ pedem potência + cadeia $(n\,g^{n-1}g')$; para produtos e quocientes, aplique exatamente as fórmulas acima.

## Exercício 3 — Latência média por usuário (regra do quociente)

**Contexto.** A latência média de uma API com $n$ usuários concorrentes é

$$
L(n)=\frac{a+bn}{1+cn},\quad a=100\text{ ms},\ b=10,\ c=0{,}05.
$$

**Enunciado.**
* (a) Derive $L'(n)$ (regra do quociente).
* (b) Avalie $L'(10)$ e interprete o sinal: a latência aumenta ou diminui ao adicionar mais um usuário no ponto $n=10$?

```python
import sympy as sp

n = sp.symbols('n', positive=True)
a, b, c = 100, 10, 0.05
L = (a + b*n)/(1+c*n)

print("L'(n) =", sp.simplify(dL))
print("L'(10) =", float(dL.subs({n: 10})))
```

```python
# Funções numéricas
fL  = sp.lambdify(n, L,  'numpy')
fdL = sp.lambdify(n, dL, 'numpy')

# Ponto de interesse
n0 = 10.0
L0 = float(L.subs({n: n0}))
slope = float(dL.subs({n: n0}))

# Faixa de n para visualização
n_vals = np.linspace(0, 60, 400)
L_vals = fL(n_vals)

# 1) Gráfico de L(n) com o ponto n=10 e a reta tangente
tangent = L0 + slope*(n_vals - n0)

plt.figure()
plt.plot(n_vals, L_vals, label="L(n)")
plt.plot(n_vals, tangent, linestyle="--", label=f"Tangente em n={n0:.0f}")
plt.scatter([n0], [L0], label=f"Ponto (n={n0:.0f}, L={L0:.2f})")
plt.axvline(n0, linestyle=":")
plt.xlabel("n (usuários concorrentes)")
plt.ylabel("L(n) (ms)")
plt.title("Latência média L(n) e tangente em n=10")
plt.legend()
plt.tight_layout()
plt.show()

# 2) Gráfico de L'(n) com destaque em n=10
d_vals = fdL(n_vals)

plt.figure()
plt.plot(n_vals, d_vals, label="L'(n)")
plt.axhline(0, linestyle="--")
plt.axvline(n0, linestyle=":")
plt.scatter([n0], [slope], label=f"L'(10) ≈ {slope:.2f}")
plt.xlabel("n (usuários concorrentes)")
plt.ylabel("L'(n) (ms por usuário)")
plt.title("Derivada L'(n) com destaque em n=10")
plt.legend()
plt.tight_layout()
plt.show()

print("Arquivos salvos: /mnt/data/Ln_tangente.png e /mnt/data/Ln_derivada.png")
```

## Exercício 5 — Crescimento de usuários (cadeia em função logística)

**Contexto.** O total de usuários de um serviço cresce logisticamente:

$$
U(t)=\frac{K}{1+A\,e^{-k t}},\quad K=10^6,\ A=9,\ k=0{,}7.
$$

Começa em $\frac{K}{1+A}$ e tende a $K$.

**Enunciado.**
* (a) Derive $U'(t)$ (escreva $U=K\,(1+A e^{-k t})^{-1}$ e aplique potência + cadeia).
* (b) Avalie $U'(5)$ e interprete como “novos usuários por dia” no dia $t=5$.

```python
import sympy as sp

K, A, k = 10**6, 9, 0.7
t = sp.symbols('t', positive=True)
U = K/1+A*sp.exp(-k*t)
dU = sp.diff(U, t)

print("U'(t) =", sp.simplify(dU))
print("U'(5) =", float(dU.subs({t: 5})))
```

```python
fU  = sp.lambdify(t, U,  'numpy')
fdU = sp.lambdify(t, dU, 'numpy')

t0   = 5.0
U0   = float(U.subs({t: t0}))
slope = float(dU.subs({t: t0}))

# Faixa para visualização
t_vals = np.linspace(0, 20, 400)
U_vals = fU(t_vals)
dU_vals = fdU(t_vals)

# 1) Gráfico de U(t) com o ponto em t=5 e reta tangente
tangent = U0 + slope*(t_vals - t0)

plt.figure()
plt.plot(t_vals, U_vals, label="U(t)")
plt.plot(t_vals, tangent, linestyle="--", label=f"Tangente em t={t0:.0f}")
plt.scatter([t0], [U0], label=f"Ponto (t={t0:.0f}, U≈{U0:.0f})")
plt.axvline(t0, linestyle=":")
plt.xlabel("t (dias)")
plt.ylabel("U(t) (usuários)")
plt.title("Crescimento logístico U(t) e tangente em t=5")
plt.legend()
plt.tight_layout()
plt.show()

# 2) Gráfico de U'(t) com destaque em t=5
plt.figure()
plt.plot(t_vals, dU_vals, label="U'(t)")
plt.axhline(0, linestyle="--")
plt.axvline(t0, linestyle=":")
plt.scatter([t0], [slope], label=f"U'(5) ≈ {slope:.0f}")
plt.xlabel("t (dias)")
plt.ylabel("U'(t) (novos usuários/dia)")
plt.title("Derivada U'(t) com destaque em t=5")
plt.legend()
plt.tight_layout()
plt.show()
```

* $U'(t)=\dfrac{K A k\,e^{-k t}}{(1+A e^{-k t})^2}$.
* Em $t=5$, $U'(5)\approx 117{,}622$ novos usuários/dia (marcado no gráfico).
* A curva de $U'(t)$ mostra o **pico de crescimento** (máxima inclinação de $U$) antes de $t=5$; depois, o crescimento desacelera à medida que $U(t)$ se aproxima do limite $K=10^6$.
